```python
import numpy as np, tensorflow as tf
from keras.models import load_model
from sklearn.cluster import KMeans

# 1. Загрузка модели
print("Загрузка модели...")
model = load_model('my_cnn_model.h5')
states = [l.strip() for l in open('stateabbr.txt') if l.strip()]

# 2. Подготовка тестовых данных
print("Подготовка данных...")
raw = [l.strip().split(',') for l in open('plants.data', encoding='latin-1') if l.strip()]
X_test = np.array([np.isin(states, r[1:]).astype('float32') for r in raw])
n_features = X_test.shape[1]
h = int(np.ceil(np.sqrt(n_features)))
w = int(np.ceil(n_features / h))
X_test = np.pad(X_test, ((0, 0), (0, h * w - n_features)), constant_values=0).reshape(-1, h, w, 1)
y_test = KMeans(n_clusters=10, n_init=10, random_state=42).fit_predict(X_test.reshape(-1, h * w))

# 3. Проверка точности
print("\nПроверка точности:")
_, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Точность модели: {acc * 100:.2f}%")

# 4. Проверка на одном примере
print("\nПроверка на одном растении:")
i = 42
pred = np.argmax(model.predict(np.expand_dims(X_test[i], axis=0), verbose=0))
print(f"Настоящий кластер: {y_test[i]} | Предсказано: {pred} | {'Угадал!' if pred == y_test[i] else 'Ошибся'}")
```
