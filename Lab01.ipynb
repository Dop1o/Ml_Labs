**Давайте добавим новый признак.**

```python
def age_category(age):
    '''
    < 30 -> 1
    >= 30, <55 -> 2
    >= 55 -> 3
    '''
    if age < 30:
        return 1
    elif age < 55:
        return 2
    elif age >= 55:
        return 3
```

```python
age_categories = [age_category(age) for age in data.Age]
data['Age_category'] = age_categories
```

**Другой способ сделать это с `apply`.**

```python
data['Age_category'] = data['Age'].apply(age_category)
```

** 4. Правда ли, что средний возраст выживших людей выше, чем у пассажиров, которые в конечном итоге умерли? **
- Да
- Нет


```python
# You code here
import pandas as pd
data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
data.head()
survived_passengers = data[data['Survived'] == 1]
death_passengers = data[data['Survived'] == 0]
average_age_survived = survived_passengers['Age'].mean()
average_age_death = death_passengers['Age'].mean()
if (average_age_survived > average_age_death):
    print("Да")
else:
    print("Нет")
```

**5. Это правда, что пассажиры моложе 30 лет. выжили чаще, чем те, кому больше 60 лет. Каковы доли выживших людей среди молодых и пожилых людей?**
- 22,7% среди молодежи и 40,6% среди пожилых
- 40,6% среди молодежи и 22,7% среди пожилых
- 35,3% среди молодежи и 27,4% среди пожилых
- 27,4% среди молодежи и 35,3% среди пожилых

```python
# You code here
import pandas as pd
data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
data.head()

survived_passengers= data[data['Survived'] == 1]
survived_young = (survived_passengers['Age'] < 30).sum()
survived_old = (survived_passengers['Age'] > 60).sum()
total_young = (data['Age'] < 30).sum()
total_old = (data['Age'] > 60).sum()

if (survived_young > survived_old):
    print("Да")
else:
    print("Нет")

prop_young = (survived_young / total_young) * 100
prop_old = (survived_old / total_old) * 100

print(f"{prop_young:.1f}% среди молодежи и {prop_old:.1f}% среди пожилых")
```

**6. Правда ли, что женщины выживали чаще мужчин? Каковы доли выживших людей среди мужчин и женщин?**
- 30,2% среди мужчин и 46,2% среди женщин
- 35,7% среди мужчин и 74,2% среди женщин
- 21,1% среди мужчин и 46,2% среди женщин
- 18,9% среди мужчин и 74,2% среди женщин

```python
# You code here
import pandas as pd
data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
data.head()
survived_passengers= data[data['Survived'] == 1]
survived_female = (survived_passengers['Sex'] == 'female').sum()
survived_male = (survived_passengers['Sex'] == 'male').sum()
total_female = (data['Sex'] == 'female').sum()
total_male = (data['Sex'] == 'male').sum()

dol_male = (survived_male / total_male) * 100
dol_female = (survived_female / total_female) * 100
if (survived_female > survived_male):
    print("Да")
else:
    print("Нет")

print(f"{dol_male:.1f}% среди мужчин и {dol_female:.1f}% среди женщин")
```
